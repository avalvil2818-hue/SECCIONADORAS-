```python
import yt_dlp
import imageio_ffmpeg
import os

# URL de la playlist
url = "https://youtube.com/playlist?list=PLKpX1atToSEzKsuW0cLkMgrrKtK89Dkat"

# Carpeta donde se guardarán TODOS los MP3
carpeta = r"D:\ESTE EQUIPO NO BORRAR\Musica\Playlist"

# Crear la carpeta si no existe
os.makedirs(carpeta, exist_ok=True)

# Localizar FFmpeg automáticamente
ffmpeg = imageio_ffmpeg.get_ffmpeg_exe()

# Configuración
opciones = {
    # Nombre de los archivos:
    # 001 - Canción.mp3
    # 002 - Canción.mp3
    # 003 - Canción.mp3
    "outtmpl": os.path.join(
        carpeta,
        "%(playlist_index)03d - %(title)s.%(ext)s"
    ),

    # Descargar el mejor audio disponible
    "format": "bestaudio/best",

    # Usar FFmpeg instalado mediante imageio-ffmpeg
    "ffmpeg_location": ffmpeg,

    # Si un video está eliminado, privado o no disponible,
    # lo salta y continúa con el siguiente
    "ignoreerrors": True,

    # Convertir el audio a MP3
    "postprocessors": [
        {
            "key": "FFmpegExtractAudio",
            "preferredcodec": "mp3",
            "preferredquality": "192",
        }
    ],

    # No volver a descargar archivos que ya existen
    "overwrites": False,
}

# Descargar playlist
with yt_dlp.YoutubeDL(opciones) as ydl:
    ydl.download([url])

print("\n======================================")
print("DESCARGA TERMINADA")
print("======================================")
print(f"Los archivos están en:\n{carpeta}")
```
